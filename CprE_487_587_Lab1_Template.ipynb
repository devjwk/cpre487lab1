```python
# This cell imports our dataset.

# Original Source: https://github.com/ksachdeva/tiny-imagenet-tfds
# Class Version Source: https://github.com/duweisu/tiny-imagenet-tfds
# Setup our dataset
# ---------------------------------------------------------

tiny_imagenet_builder = TinyImagenetDataset()

# this call (download_and_prepare) will trigger the download of the dataset
# and preparation (conversion to tfrecords)
#
# This will be done only once and on next usage tfds will
# use the cached version on your host.
tiny_imagenet_builder.download_and_prepare(download_dir="~/tensorflow-datasets/downloads")

# class_names = tiny_imagenet_builder.info.features['label'].names
ds = tiny_imagenet_builder.as_dataset()
ds_train, ds_val = ds["train"], ds["validation"]
assert(isinstance(ds_train, tf.data.Dataset))
assert(isinstance(ds_val, tf.data.Dataset))

# Training Dataset
ds_train = ds_train.shuffle(1024).prefetch(tf.data.AUTOTUNE)

# Validation Dataset
ds_val = ds_val.shuffle(1024).prefetch(tf.data.AUTOTUNE)

# Dataset metadata
ds_info = tiny_imagenet_builder.info
```

```python
# We need to read the "human readable" labels so we can translate with the numeric values
# Read the labels file (words.txt)
with open(os.path.abspath('wnids.txt'), 'r') as f:
    wnids = [x.strip() for x in f]

# Map wnids to integer labels
wnid_to_label = {wnid: i for i, wnid in enumerate(wnids)}
label_to_wnid = {v: k for k, v in wnid_to_label.items()}

# Use words.txt to get names for each class
with open(os.path.abspath('words.txt'), 'r') as f:
    wnid_to_words = dict(line.split('\t') for line in f)
    for wnid, words in wnid_to_words.items():
        wnid_to_words[wnid] = [w.strip() for w in words.split(',')]
        
class_names = [str(wnid_to_words[wnid]) for wnid in wnids]
```

```python
# Helper function to get the label name
def img_class(img_data, idx=None):
    image, label, id, label_name = img_data["image"], img_data["label"], img_data["id"], img_data["metadata"]['label_name']
    # Handle batches of images correctly
    if idx != None:
        image, label, id, label_name = img_data["image"][idx], img_data["label"][idx], img_data["id"][idx], img_data["metadata"]['label_name'][idx]
    
    return f"{label_name} (class index: {label} - id: {id})"


# Helper function to show basic info about an image
def img_info(img, idx=None, display=True, title_apend=""):
    image = img['image']

    # Print the class
    class_str = img_class(img, idx)
    print(f"Label: {class_str}")
    
    # Display the image
    if display:
        plt.figure()
        plt.title(title_apend + class_str)
        # Handle batches correctly
        if image.shape.ndims > 3:
            plt.imshow(image.numpy().reshape(64, 64, 3))
        else:
            plt.imshow(image.numpy())
```

```python
# Running infrence on our model
# We can run an infrence of our model by doing the following (we are doing batches of 1 here)
for example in ds_train.batch(1).take(1):
    img_info(example)
    
    # Make a prediction
    pred = model.predict(tf.cast(example["image"], tf.float32)/255.0)
    # print(f'Raw 200 Class Weighted Prediction:\n{pred}') # Uncomment to see the raw prediction
    
    # What is out best guess?
    best_guess = tf.math.argmax(pred, axis=1).numpy() # Our output is 200 weighted value, we want the most likely
    print(f'Best Guess [class index]: {class_names[best_guess[0]]} [{best_guess[0]}]')
    print(f'Best Guess Confidence (percent / 1.0): {pred[0][best_guess]}')

    # What are our top 15 guesses?
    top_15 = tf.math.top_k(pred, k=15)
    print(f'Top 15 Guesses (class index): {[f"{class_names[idx]} [{idx}]" for idx in top_15.indices[0]]}')
    print(f'Top 15 Guesses Confidence (percent / 1.0): {top_15.values}')
```

```python
# TODO: Calculate the Top-1, Top-5, and Top-10 Accuracy of the validation dataset
total = acc_top1 = acc_top5 = acc_top10 = 0

# TODO: Your Code Here
# Batched inference over the first 10,000 validation images (the whole validation split)
for batch in ds_val.take(10000).batch(100):
    pred = model.predict(tf.cast(batch["image"], tf.float32) / 255.0, verbose=0)
    labels = batch["label"].numpy()

    # hits[i, k] is True when the k-th most likely class of image i is the ground truth
    top_10 = tf.math.top_k(pred, k=10).indices.numpy()
    hits = top_10 == labels[:, None]

    acc_top1 += int(hits[:, :1].sum())
    acc_top5 += int(hits[:, :5].sum())
    acc_top10 += int(hits.sum())
    total += len(labels)

print(f'Total Images: {total}')
print(f'Top-1 Accuracy: {acc_top1 / total * 100:.2f}%')
print(f'Top-5 Accuracy: {acc_top5 / total * 100:.2f}%')
print(f'Top-10 Accuracy: {acc_top10 / total * 100:.2f}%')
```

```python
# We can view the layer weights. Here we consider them as images of feature filters applied to intermediate feature map images.
# TODO: Visualize the 2 convolutional layers filter sets (weights) (one at the beginning and one at the end)

# TODO: Your Code Here
conv_layers = [layer for layer in model.layers if isinstance(layer, tf.keras.layers.Conv2D)]
early_layer, late_layer = conv_layers[0], conv_layers[-1]

def show_filters(layer, num_filters=8):
    kernel, bias = layer.get_weights()
    print(f'--- {layer.name} ---')
    print(f'Data type: {kernel.dtype}')
    print(f'Kernel dimensions (H, W, in, out): {kernel.shape}, bias dimensions: {bias.shape}')
    print(f'Memory: {kernel.nbytes} (kernel) + {bias.nbytes} (bias) = {kernel.nbytes + bias.nbytes} bytes')

    fig, axes = plt.subplots(1, num_filters, figsize=(2 * num_filters, 2.6))
    for i, ax in enumerate(axes):
        filt = kernel[:, :, :, i]
        if filt.shape[-1] == 3:
            # RGB input: show the whole filter as a color image, rescaled to [0, 1]
            ax.imshow((filt - filt.min()) / (filt.max() - filt.min() + 1e-8))
        else:
            # Many input channels: show the 2D slice that is applied to input channel 0
            ax.imshow(filt[:, :, 0], cmap='gray')
        ax.set_title(f'filter {i}')
        ax.axis('off')
    fig.suptitle(f'Weights of {layer.name} {kernel.shape}')
    plt.tight_layout()
    plt.show()

show_filters(early_layer)
show_filters(late_layer)
```

```python
# We can view the layer outputs as well. Here we consider them as images of the spatial location of features.
# TODO: Visualize the 2 convolutional layers outputs (intermediate feature maps) (one at the beginning and one at the end)

# TODO: Your Code Here
# One model that returns the output of every layer (reused by the export cell below)
feature_model = tf.keras.Model(inputs=model.inputs, outputs=[layer.output for layer in model.layers])

def run_all_layers(img_data):
    input_tensor = tf.expand_dims(tf.cast(img_data["image"], tf.float32) / 255.0, axis=0)
    outputs = feature_model(input_tensor, training=False)
    return {layer.name: output.numpy() for layer, output in zip(model.layers, outputs)}

def show_feature_maps(features, layer_name, num_maps=8):
    print(f'--- {layer_name} ---')
    print(f'Data type: {features.dtype}')
    print(f'Feature map dimensions (N, H, W, C): {features.shape}')
    print(f'Memory: {features.nbytes} bytes')

    fig, axes = plt.subplots(1, num_maps, figsize=(2 * num_maps, 2.6))
    for i, ax in enumerate(axes):
        ax.imshow(features[0, :, :, i], cmap='viridis')
        ax.set_title(f'channel {i}')
        ax.axis('off')
    fig.suptitle(f'Feature maps of {layer_name} {features.shape}')
    plt.tight_layout()
    plt.show()

# Use sample image 0 as the test input
img_info(sample_imgs[0], title_apend='Input: ')
plt.show()

sample_outputs = run_all_layers(sample_imgs[0])
show_feature_maps(sample_outputs[early_layer.name], early_layer.name)
show_feature_maps(sample_outputs[late_layer.name], late_layer.name)
```

```python
# TODO: Sample Profiling - Inference for a single image:
import time

def preprocess(images):
    # uint8 [0, 255] -> float32 [0, 1]; a single image also gets the batch dimension
    images = tf.cast(images, tf.float32) / 255.0
    return tf.expand_dims(images, axis=0) if images.shape.ndims == 3 else images

# Warm-up: the first predict call builds the inference graph, which is not part of the inference itself
model.predict(preprocess(sample_imgs[0]["image"]), verbose=0)

# Perform the inference profiling (one profiled inference for each of the 3 sample images):
single_latency_ms = []
# Starts Profile logging
tf.profiler.experimental.start(log_dir_run)
try:
    for index, img_data in enumerate(sample_imgs):
        with tf.profiler.experimental.Trace('single_inference', step_num=index, _r=1):
            # Actual inference
            # TODO: Your Code Here
            start = time.perf_counter()
            model.predict(preprocess(img_data["image"]), verbose=0)
            single_latency_ms.append((time.perf_counter() - start) * 1000)
finally:
    tf.profiler.experimental.stop()

# The profiler adds overhead, so also report the median of 20 un-profiled runs per image
print(f'{"Image":<7}{"Class":<18}{"Profiled (ms)":<15}Median of 20 runs (ms)')
for index, img_data in enumerate(sample_imgs):
    repeats = []
    for _ in range(20):
        start = time.perf_counter()
        model.predict(preprocess(img_data["image"]), verbose=0)
        repeats.append((time.perf_counter() - start) * 1000)
    name = wnid_to_words[wnids[int(img_data["label"].numpy())]][0]
    print(f'{index:<7}{name:<18}{single_latency_ms[index]:<15.2f}{np.median(repeats):.2f}')
```

```python
# Layer-wise computation latency plot (computation time on the y-axis, layer name on the x-axis)
# The values are "Avg. self-time (us)" read from TensorBoard > Profile > framework_op_stats (this is the view the
# lab manual calls tensorflow_stats). Screenshots of the same tables are in the report appendix.
#   online: run 2026_10_07_16_14_51 = 1000 single-image predict calls (1000 occurrences of every operation)
#   batch : run 2026_10_07_16_16_16 = 5 predict calls with batch size 200 (5 occurrences), divided by 200 below
# The operation names carry an extra suffix: layer conv2d is 'sequential_1/conv2d_1/Relu', layer conv2d_1 is
# 'sequential_1/conv2d_1_2/Relu', layer dense is 'dense_1/Relu' and layer dense_1 is 'dense_1_2/BiasAdd' + 'Softmax'.
# flatten does no computation and has no entry. These numbers must be re-read if the profiling cells are re-run.
layer_names = ['conv2d', 'conv2d_1', 'max_pooling2d', 'conv2d_2', 'conv2d_3', 'max_pooling2d_1',
               'conv2d_4', 'conv2d_5', 'max_pooling2d_2', 'dense', 'dense_1']
assert layer_names == [layer.name for layer in model.layers if layer.name != 'flatten']

online_us = np.array([283, 538, 93, 167, 187, 59, 90, 94, 46, 180, 30 + 10])
# max_pooling2d_2 and dense_1 were below the visible rows; they are the per-type totals minus the visible rows
batch200_us = np.array([8990, 62067, 4213, 12552, 17453, 1252, 3454, 5426, 485, 733, 180 + 59]) / 200

def mac_count(layer):
    # Multiply-accumulate operations needed for one image (0 for pooling layers)
    if not layer.weights:
        return 0
    kernel = layer.get_weights()[0]
    if kernel.ndim == 4:  # conv: every output value needs kernel_h * kernel_w * in_channels MACs
        return int(np.prod(layer.output.shape[1:]) * np.prod(kernel.shape[:3]))
    return int(kernel.size)  # dense: in * out

macs = np.array([mac_count(model.get_layer(name)) for name in layer_names])

print(f'{"Layer":<18}{"online (us)":>12}{"share":>8}{"batch 200 (us)":>16}{"share":>8}{"MACs (M)":>10}')
for name, online, batch, mac in zip(layer_names, online_us, batch200_us, macs):
    print(f'{name:<18}{online:>12.0f}{online / online_us.sum() * 100:>7.1f}%'
          f'{batch:>16.1f}{batch / batch200_us.sum() * 100:>7.1f}%{mac / 1e6:>10.2f}')
print(f'{"total":<18}{online_us.sum():>12.0f}{"":>8}{batch200_us.sum():>16.1f}{"":>8}{macs.sum() / 1e6:>10.2f}')

fig, axes = plt.subplots(1, 2, figsize=(16, 5))
for ax, values, title in ((axes[0], online_us, 'Online inference (batch size 1)'),
                          (axes[1], batch200_us, 'Batch inference (batch size 200)')):
    ax.bar(layer_names, values)
    ax.set_xlabel('Layer name')
    ax.set_ylabel('Computation time per image (us)')
    ax.set_title(f'Layer-wise computation latency, CPU - {title}')
    plt.setp(ax.get_xticklabels(), rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# TODO: Sample Profiling - Online Inference:

# The same 1,000 validation images are used for the online and the batch experiments.
# They are loaded into memory first so that reading the dataset is not counted as inference time.
profile_images = tf.stack([example["image"] for example in ds["validation"].take(1000)])

# Vary this from 10, 100, 1000 to simulate multiple online inference
loop_index = [10, 100, 1000]

online_results = {}
for idx in loop_index:
    # Starts Profile logging
    tf.profiler.experimental.start(log_dir_run)

    # Actual online inference
    # TODO: Your Code Here
    try:
        start = time.perf_counter()
        for image in profile_images[:idx]:
            model.predict(preprocess(image), verbose=0)
        latency = time.perf_counter() - start
    finally:
        tf.profiler.experimental.stop()

    online_results[idx] = latency
    print(f'{idx:>5} images: end-to-end latency {latency:8.3f} s, '
          f'throughput {idx / latency:8.2f} images/s, {latency / idx * 1000:7.2f} ms/image')
```
